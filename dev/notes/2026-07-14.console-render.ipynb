Console Render
==============================================================================


```python
from rich.console import Console
from rich.style import Style
from rich.text import Text

text = Text.assemble(
    *(
        (s, s)
        for s in (
            "red",
            "bold blue",
            "pretty",
            "repr.str",
        )
    ),
    ("hex color", Style(color="#AABBCC")),
)


def render(renderable, **console_kwds):
    return list(Console(**console_kwds).render(renderable))


render(text)
```

```python
render(text, no_color=True)
```

```python
render(text, no_color=True, force_terminal=False, color_system="standard")
```
